## STEP 3 : MENCARI OUTLIER

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Daftar kolom numerik yang mau dicek
features = ['Daily Time Spent on Site', 'Age', 'Area Income', 'Daily Internet Usage']

plt.figure(figsize=(12, 8))
for i, col in enumerate(features):
    plt.subplot(2, 2, i+1)
    sns.boxplot(x=df[col])
    plt.title(f'Outlier Check: {col}')

plt.tight_layout()
plt.show()
```

nah ini objektifnya buat liat anomali/kebobrokan/keburukan dari data yang kita punnya. di sini ada anomali pada kolom "Area Income".

### ngebenerin outlier di area income (anomali)

```python
#itung statistik dasar
Q1 = df['Area Income'].quantile(0.25)
Q3 = df['Area Income'].quantile(0.75)
IQR = Q3 - Q1

print(f"Kuartil 1 (Q1): {Q1}")
print(f"Kuartil 3 (Q3): {Q3}")
print(f"Rentang IQR   : {IQR}")
```

```python
#nentuin batas aman
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\nBatas Bawah (Pagar Luar): {lower_bound}")
print(f"Batas Atas  (Pagar Luar): {upper_bound}")
```

```python
#intip baris anomali
outliers = df[(df['Area Income'] < lower_bound) | (df['Area Income'] > upper_bound)]
print(f"\nJumlah data yang dianggap outlier: {len(outliers)} baris")
print("Contoh data outlier (pendapatan terlalu rendah):")
print(outliers[['Area Income']].head())
```

```python
#eksekusi
jumlah_awal = df.shape[0]
df = df[(df['Area Income'] >= lower_bound) & (df['Area Income'] <= upper_bound)]
jumlah_akhir = df.shape[0]

print(f"\n--- HASIL AKHIR ---")
print(f"Data awal : {jumlah_awal}")
print(f"Data akhir: {jumlah_akhir}")
print(f"Total yang dihapus: {jumlah_awal - jumlah_akhir}")
```

### 2. Model Training
Kita menggunakan algoritma **Logistic Regression**, yang merupakan standar untuk masalah klasifikasi biner (seperti klik vs tidak klik). Model akan belajar pola dari `X_train` untuk memprediksi target `y_train`.

### 3. Model Evaluation
Evaluasi dilakukan untuk melihat sejauh mana performa model dalam memprediksi data testing. Kita menggunakan **Classification Report** untuk melihat nilai akurasi dan **Confusion Matrix** sebagai visualisasi untuk membandingkan jumlah prediksi yang benar dan yang salah secara mendetail.

```python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

y_pred = model.predict(X_test)

print('--- Classification Report ---')
print(classification_report(y_test, y_pred, zero_division=0))

plt.figure(figsize=(6,4))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Tidak Klik', 'Klik'],
            yticklabels=['Tidak Klik', 'Klik'])
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.title('Confusion Matrix - Evaluasi Prediksi Klik Iklan')
plt.show()
```
